## solar workflow

```python
# define necessary files
era5_path = os.path.join(data_basefolder, 'weather_data/ERA5')
gsa_ghi = os.path.join(data_basefolder, 'weather_data/global_solar_atlas/GHI_NER5.1.tif')
gsa_dni = os.path.join(data_basefolder, 'weather_data/global_solar_atlas/DNI_NER5.1.tif')
```

```python
# read solar placements
results_folder2 = "/storage_cluster/projects/2021_s-chen_PhD/projects_root/0013_IMP_master_students_teach_repo/solar_wind_placements/"
solar_placements = gk.vector.extractFeatures(os.path.join(results_folder2, 'placements_solar.shp'))
# define the solar plant parameters for these placements
solar_placements["capacity"] = 2000

```

### call RESKit to simulate solar power

```python

gen = rk.solar.openfield_pv_era5(
        placements=solar_placements,
        era5_path=era5_path,
        global_solar_atlas_ghi_path=gsa_ghi,
        global_solar_atlas_dni_path=gsa_dni,
        module="WINAICO WSx-240P6",
        elev=300,
        tracking="fixed",
        inverter=None,
        inverter_kwargs={},
        tracking_args={},
        output_netcdf_path=None,
        output_variables=None,
    )
gen
```
